```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.style as style
```

```python
red_wine = pd.read_csv('winequality-red.csv',sep=";")
white_wine = pd.read_csv('winequality-white.csv',sep=";")
```

```python
red_corr = red_wine.corr()['quality'][:-1].sort_values(ascending=False)
```

```python
white_corr = white_wine.corr()['quality'][:-1].sort_values(ascending=False)
```

```python
style.use('fivethirtyeight')
fig,ax = plt.subplots(figsize=(9,5))
ax.barh(white_corr.index,white_corr,left=2,height=0.5)
ax.barh(red_corr.index,red_corr,height=0.5)
ax.grid(visible=False)
ax.set_xticks([])
ax.set_yticks([])
plt.show()
```

```python
style.use('fivethirtyeight')
fig, ax = plt.subplots(figsize=(9, 5))
ax.barh(white_corr.index, white_corr, left=2, height=0.5)
ax.barh(red_corr.index, red_corr, height=0.5, left=-0.1)

ax.grid(False)
ax.set_yticklabels([])
ax.set_xticklabels([])

x_coords = {'Alcohol': 0.82, 'Sulphates': 0.77, 'pH': 0.91,
            'Density': 0.80, 'Total Sulfur Dioxide': 0.59,
            'Free Sulfur Dioxide': 0.6, 'Chlorides': 0.77,
            'Residual Sugar': 0.67, 'Citric Acid': 0.76,
            'Volatile Acidity': 0.67, 'Fixed Acidity': 0.71}
y_coord = 9.8
for y_label, x_coord in x_coords.items():
    ax.text(x_coord, y_coord, y_label)
    y_coord -= 1

ax.axvline(0.5,c='grey',alpha=0.1,linewidth=1,ymin=0.1,ymax=0.9)

ax.axvline(1.45,c='grey',alpha=0.1,linewidth=1,ymin=0.1,ymax=0.9)

plt.show()
```
